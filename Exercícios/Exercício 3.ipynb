**QUESTÃO 1: Crie uma lista preenchida com os 5 primeiros colocados de um Campeonato de Futebol na ordem de colocação. Depois mostre:**

 *Os últimos 2 colocados;*

 *Em que posição da tabela se encontra o Barcelona.*

QUESTÃO 3: Faça um programa que leia o nome e a média de um aluno e guarde-os em um dicionário. Em seguida, a partir da média (para ser aprovado deve ter média >=50), gere a situação final do aluno (‘AP’ ou ‘RP’), que também deve ser guardada neste dicionário. No final, mostre todo o conteúdo deste dicionário;

```python
aluno = {}

aluno['nome'] = input('Digite o nome do aluno: ')
aluno['media'] = float(input(f'Digite a média de {aluno['nome']}: '))

if aluno['media'] >= 50:
    aluno['situacao'] = 'AP'

else:
    aluno['situacao'] = 'RP'

print('\nDicionário: ')
print(aluno)
```

QUESTÃO 4: Faça um programa que leia o nome e peso de 3 pessoas e no final mostre o nome da pessoa mais pesada e a mais leve;

```python
pessoa = {}

for i in range(1, 4):
    nome = input(f'Digite o nome da {i} pessoa: ')
    peso = float(input(f'Digite o peso de {nome}: '))
    pessoa[nome] = peso

mais_pesado = max(pessoa, key=pessoa.get)
mais_leve = min(pessoa, key=pessoa.get)

print('\n Resultado')
print(f'Pessoa mais pesada: {mais_pesado} ({pessoa[mais_pesado]} kg)')
print(f'Pessoa mais leve: {mais_leve} ({pessoa[mais_leve]} kg)')

```

QUESTÃO 5: Desenvolva um programa que leia o nome, idade e sexo de n pessoas. No final, mostre:
 A média de idade do grupo;
 Quantas mulheres têm menos de 20 anos.
Dica: Em Python, os operadores booleanos básicos são and, or e not.

```python
n = int(input("Digite quantas pessoas estará no programa: "))

pessoas = []

for i in range (1, n + 1):
 print(f'Dados da {i} pessoa: ')
 nome = input('Nome: ')
 idade = int(input('Idade: '))
 sexo = input('Sexo[M or F]: ').strip().upper()

 pessoa = {'nome': nome, 'idade': idade, 'sexo': sexo}
 pessoas.append(pessoa)

soma_idades = 0
mulheres_menos_20 = 0

for p in pessoas:
 soma_idades += p['idade']

 if p['sexo'] == 'F' and p['idade'] < 20:
  mulheres_menos_20 += 1

media_idade = soma_idades / n

print('\n Resultado')
print(f"Total de pessoas cadastradas: {n}")
print(f"Média de idade do grupo: {media_idade:.1f} anos")
print(f"Mulheres com menos de 20 anos: {mulheres_menos_20}")
```

QUESTÃO 7: Usando uma lista de ingredientes já preenchida para uma receita de bolo. Crie
dois conjuntos representando os ingredientes que duas pessoas diferentes já têm em casa.
Mostre quais ingredientes da receita ainda faltam comprar.
Dica: Use diferença de conjuntos entre a lista de ingredientes e a união dos dois conjuntos

```python
receita = ['fermento', 'farinha de trigo', 'acucar', 'leite', 'ovo', 'cenoura']

pessoa1 = {'fermento', 'farinha de trigo', 'acucar'}
pessoa2 = {'leite', 'ovo'}

ingredientes = set(receita)

ingredientes_totais = pessoa1 | pessoa2

faltante = ingredientes - ingredientes_totais
print(ingredientes_totais)
print(faltante)
```

QUESTÃO 8: Crie um dicionário para armazenar os dados de um produto (nome, preço e
quantidade em estoque). Peça ao usuário os dados de 3 produtos diferentes e guarde cada
dicionário em uma lista. No final, percorra a lista e mostre, para cada produto, seu nome e o
valor total em estoque (preço × quantidade).

```python
produtos = []

for i in range(1, 4):
 nome = input('Nome do produto {i}: ')
 preco = float(input('Preço do produto {i}: '))
 qtd_estoque = int(input('Quantidade do produto {i} em estoque: '))

 produto = {
  'nome': nome,
  'preco': preco,
  'quantidade': qtd_estoque
 }

 produtos.append(produto)

print('\n -- Relatório do Estoque --')
for p in produtos:
 total_produto = p['preco'] * p['quantidade']
 print(f'Produto: {p['nome']} | Valor total em estoque: R$ {total_produto: .2f}')

```
