QUESTÃO 1: Apresente a porcentagem de missões que deram certo.

```python
import numpy as np

dataset = np.loadtxt('Inserts/space.csv', delimiter=';', dtype=str, encoding='utf-8')

filter = dataset[:, -1]

print(filter)

porcentagem_sucesso = np.mean(filter == 'Success') * 100

print(f'Missões que foram um sucesso: {porcentagem_sucesso:2f}')
```

QUESTÃO 2: Qual a média de gastos de uma missão espacial se baseando em missões que possuam valores disponíveis (> 0)?

```python
dataset = np.loadtxt('Inserts/space.csv', delimiter=';', dtype=str, encoding='utf-8')

filter = dataset[1:, -2]

# 2. Remove strings vazias
filter = filter[filter != '']

# 3. Converte para float
custos = filter.astype(float)

# 4. Filtra valores maiores que zero
custos_validos = custos[custos > 0]

# 5. Média dos custos
media_custos = np.mean(custos_validos)

print(f'Média dos custos (> 0): {media_custos:.2f}')
```

QUESTÃO 3: Encontre quantas missões espaciais neste Dataset foram realizadas pelos Estados Unidos (EUA)

```python
filter = dataset[1:, 2]

usa = np.char.find(filter, 'USA') >= 0

quantidade_missoes = np.sum(usa)

print(f'{quantidade_missoes} missões foram realizadas pelos USA.')
```

QUESTÃO 4: Encontre qual foi a missão mais cara realizada pela empresa “SpaceX”

```python
# 2. Ignora o cabeçalho dos dados
dados = dataset[1:]

# 3. Filtra apenas as linhas da SpaceX
dados_spacex = dados[dados[:, 1] == 'SpaceX']

# 4. Extrai a coluna de custos e converte para float
# Se houver custos vazios, você pode substituir temporariamente ou filtrar
custos_spacex = dados_spacex[:, -2].astype(float)

# 5. Encontra o índice da missão mais cara
idx_max = np.argmax(custos_spacex)

# 6. Recupera as informações da missão
num_missao = dados_spacex[idx_max, 0]
custo_max = custos_spacex[idx_max]

print(f'Num da missão mais cara da SpaceX: {num_missao}')
print(f'Custo: ${custo_max}')
```

QUESTÃO 5: Mostre o nome das empresas que já realizaram missões espaciais, junto com suas respectivas quantidades de missões (use o for no final para mostrar as informações)

```python
filter = dataset[1:, 1]

empresas_unicas, contagem = np.unique(filter, return_counts=True)

for i in range(len(empresas_unicas)):
    print(f'Empresa: {empresas_unicas[i]:<25} | Quantidade de missões: {contagem[i]}')
```
