QUESTÃO 1: Crie dois NumPy Arrays unidimensionais de tamanho 8: um formado apenas por 1’s e outro formado por números aleatórios entre 0 e 9. Some estes dois NumPy Arrays e guarde o resultado dentro de um terceiro NumPy Array. Por fim, faça o seguinte:
 Se a soma de todos os elementos do Array resultante for >= 40, remodele este NumPy Array para se tornar uma matriz com mais linhas do que colunas. Senão, remodele para que se torne uma matriz com mais colunas do que linhas.

```python
import numpy as np
```

```python
arr1 = np.ones(8)
arr2 = np.random.randint(0, 10, 8)

array_result = arr1 + arr2

if np.sum(array_result) >= 40:
    array_result = array_result.reshape([2,-1])

else:
    array_result = array_result.reshape([-1,2])

print(f'Array 1: {arr1}')
print()
print(f'Array 2: {arr2}')
print()
print(f'O resultado da matriz é: {array_result}')

```

QUESTÃO 3: Crie uma matriz de tamanho qualquer. Extraia seu número de linhas e colunas,
multiplique-os, e diga se esta matriz poderia se tornar um vetor unidimensional com número par ou ímpar de elementos.

```python
matriz = np.random.randint(0, 10, [5,5])

soma_linhas = matriz.shape[0]
soma_colunas = matriz.shape[1]
linhas_por_colunas = soma_linhas * soma_colunas
paridade = 'par' if linhas_por_colunas % 2 == 0 else 'ímpar'

print(f'A quantidade de linhas é: {soma_linhas}')
print(f'A quantidade de colunas é: {soma_colunas}')
print(f'O total de elementos ao achatar é: {linhas_por_colunas}')
print(f'Esta matriz se tornaria um vetor unidimensional com número {paridade} de elementos.')

```

QUESTÃO 4: Crie uma matriz de tamanho 4x4 formada por números aleatórios inteiros entre 1 e 50 (use seed = 10 antes)
a) Mostre o resultado da média de cada linha e cada coluna da matriz gerada
b) Apresente o maior valor das médias das linhas e também das colunas
c) Mostre a quantidade de aparições de cada um dos números gerados na matriz. Em
seguida, mostre apenas os números que aparecem 2 vezes

```python
np.random.seed(10)
matriz = np.random.randint(1, 51, [4, 4])

print('Matriz Gerada')
print(matriz)
print()

media_linhas = matriz.mean(axis=1)
media_colunas = matriz.mean(axis=0)
                    
maior_media_linhas = media_linhas.max()
maior_media_colunas = media_colunas.max()

print('a)')
print(f'Maior média entre as linhas: {maior_media_linhas}')
print()
print('b)')
print(f'Maior media entre as colunas: {maior_media_colunas}')
```

QUESTÃO 5: Crie uma matriz de tamanho 3x3 formada apenas por zeros e outra também de
tamanho 3x3 formada apenas por uns. Some as duas matrizes elemento a elemento e
multiplique o resultado por um escalar de sua escolha. Por fim, remodele (reshape) a matriz resultante para um array unidimensional de 9 elementos.

```python
matriz0 = np.zeros([3, 3])
matriz1 = np.ones([3, 3])

soma_matrizes = matriz0 + matriz1

resultado = soma_matrizes * 7

matriz1d = resultado.reshape(9)

print(f'Matriz de 0: {matriz0}')
print(f'Matriz 1: {matriz1}')
print()
print(f'Soma das matrizes: {soma_matrizes}')
print(f'Resultado multiplicado pelo escalar 7: {resultado}')
print(f'Matriz final: {matriz1d}')
```

QUESTÃO 6: Crie um NumPy Array unidimensional com os números de 10 a 30, espaçados de 2
em 2 usando arange. Mostre o maior valor do array e seu índice, o menor valor e seu índice, e a soma de todos os elementos.

```python
arr = np.arange(10,31,2)

print(f'array: {arr}')

print(f'Maior valor do array: {arr.max()},\n Índice: {arr.argmax()}')
print(f'Menor valor do array: {arr.min()},\n Índice: {arr.argmin()}')
print()

print(f'Soma de todos os elementos: {arr.sum()}')
```

QUESTÃO 7: Crie dois NumPy Arrays unidimensionais de tamanho 5 com números aleatórios
entre 1 e 10 (use seed=7). Concatene os dois arrays em um terceiro e, transforme-o em uma matriz com 2 linhas e 5 colunas.

```python
np.random.seed(7)

arr1 = np.random.randint(1, 11, 5)
arr2 = np.random.randint(1, 11, 5)

arr3 = np.concatenate((arr1, arr2))

matriz = arr3.reshape([2,5])

print(f'Array 1: {arr1}')
print(f'Array 2: {arr2}')
print()

print(f'Array Concatenado: {arr3}')

print(f'Matriz do Array concatenado: {matriz}')

```
