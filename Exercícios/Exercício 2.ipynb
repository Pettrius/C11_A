QUESTÃO 1: Crie um programa que leia seu nome completo e mostre:
• Seu nome com todas as letras maiúsculas;
• Seu nome com todas as letras minúsculas;
• Quantas letras ao todo tem seu nome;
• E como seria se trocássemos seu último nome para “do Inatel”.

```python
nome = input('Digite seu nome completo: ')

nome_maiusculo = nome.upper()
nome_minusculo = nome.lower()
tamanho_nome = len(nome) 
ultimo_nome = nome.replace('cardoso', 'do Inatel')

print (nome_maiusculo)
print (nome_minusculo)
print (tamanho_nome)
print (ultimo_nome)
```

QUESTÃO 2: Mostre a tabuada de um número que o usuário escolher dentro de um intervalo
específico também escolhido por ele.

```python
numero = int(input('Digite um número inteiro para saber a tabuada dele: '))
inicio = int(input('Digite o início da tabuada: '))
fim = int(input('Digite o fim da tabuada: '))

for i in range (inicio, fim + 1):
    resultado = numero * i
    print(f'{numero} x {i} = {resultado}')
```

QUESTÃO 3: Faça um programa que leia o sexo de uma pessoa e diga se ela é homem (caso seja
digitado M) ou mulher (caso seja digitado F). Caso seja digitado algo inválido, continue
perguntando até que o usuário entre com um sexo válido.

```python
sexo = input('Digite seu sexo (M/F): ')

while sexo != 'M' and sexo != 'F':
    sexo = input('Digite seu sexo (M/F): ')

if sexo == 'M':
    print('Masculino')

elif sexo == 'F':
    print('Feminino')

else:
    print('Sexo inválido')
```

QUESTÃO 4: Desenvolva um script que pergunte a distância de uma viagem em Km. Calcule o preço da passagem, cobrando R$0.50 por Km para viagens até 200Km e R$0.45 para viagens mais longas.

```python
distancia = float(input('Digite a distância em Km: '))

if distancia <= 200:
    preco = distancia * 0.50
    print(f'O preço da passagem é: R${preco:.2f}')

elif distancia > 200:
    preco = distancia * 0.45
    print(f'O preço da passagem é: R${preco:.2f}')

else:
    print('Distância inválida')
```

QUESTÃO 5: Faça um programa que leia um número entre 1000 e 9999 e mostre na tela:
• Qual o número da unidade;
• Número da dezena;
• Número da centena;
• E número do milhar.

```python
numero = int(input('Digite um número inteiro entre 1000 e 9999: '))

if numero >= 1000 and numero <= 9999:
    unidade = numero % 10
    dezena = (numero // 10) % 10
    centena = (numero // 100) % 10
    milhar = (numero // 1000) % 10

    print(f'Unidade: {unidade}')
    print(f'Dezena: {dezena}')
    print(f'Centena: {centena}')
    print(f'Milhar: {milhar}')
```

QUESTÃO 6: Peça ao usuário para entrar com um número decimal. Em seguida, aplique e mostre o resultado:
• Da raiz quadrada deste número;
• Função teto;
• Função chão;
• Sua parte inteira.

```python
import math

numero = float(input('Digite um número decimal: '))

raiz_quadrada = math.sqrt(numero)

teto = math.ceil(numero)
chao = math.floor(numero)
inteiro = math.trunc(numero)

print(f'Raiz quadrada: {raiz_quadrada:.2f}')
print (f'Teto: {teto}')
print (f'Chão: {chao}')
print (f'Parte inteira: {inteiro}')
```

QUESTÃO 7: Faça um programa que leia uma palavra. Esse programa deve percorrer a palavra e
imprimir cada letra em maiúsculo, uma letra por linha. Ao final, deverá informar quantas vogais
a palavra tem e se a letra “a” está presente nela.

```python
palavra = input('Digite uma palavra: ')

total_vogais = 0
vogais = 'aeiouAEIOU'

for letra in palavra:
    letra_maiuscula = letra.upper()
    print(f'Letra: {letra_maiuscula}')

    if letra_maiuscula in vogais:
        total_vogais += 1

print(f'Total de vogais: {total_vogais}')

if 'a' in palavra:
    print('A letra "a" está presente na palavra.')
```

QUESTÃO 8: Crie um programa que leia dois números calculando as operações abaixo. Em seguida, informe ao final das operações se o resultado da adição é par ou ímpar:
• Adição
• Subtração
• Multiplicação
• Divisão
• Resto da divisão
• Potência, o primeiro número elevado ao segundo.

```python
numero1 = float(input('Digite um número inteiro: '))
numero2 = float(input('Digite outro número inteiro: '))

adicao = numero1 + numero2
subtracao = numero1 - numero2
multiplicacao = numero1 * numero2
divisao = numero1 / numero2
resto_divisao = numero1 % numero2
potencia = numero1 ** numero2

print(f'A soma dos números é: {adicao}')
print(f'A subtração dos números é: {subtracao}')
print(f'A multiplicação dos números é: {multiplicacao}')
print(f'A divisão dos números é: {divisao}')
print(f'O resto da divisão dos números é: {resto_divisao}')
print(f'O resultado da potência dos números é: {potencia}')

if (adicao % 2 == 0):
    print('A soma é par.')
else:
    print('A soma é ímpar.')

    
```
